## One devil, one record

A **dictionary** keeps related facts together, each one with a label. Curly
braces, and `label: value` pairs:


```python
gun_devil = {'name': 'Gun', 'fear': 99, 'status': 'at large'}

print(gun_devil)
```

You pull a fact out by asking for its **label**:


```python
print(gun_devil['name'])
print(gun_devil['fear'])
print(gun_devil['status'])
```

Compare that to a list. With a list you'd write `devil[0]` and have to remember
that slot 0 happens to be the name. With a dictionary you write
`devil['name']`, which still makes sense in six months.

Records change, too — devils get reclassified:


```python
gun_devil['fear'] = 100          # change an existing fact
gun_devil['weakness'] = 'none'   # add a brand new one

print(gun_devil)
```

## The whole file

One record is a dictionary. The **archive** is a list of them:


```python
devils = [
    {'name': 'Gun',      'fear': 99, 'status': 'at large'},
    {'name': 'Bomb',     'fear': 82, 'status': 'at large'},
    {'name': 'Katana',   'fear': 74, 'status': 'contained'},
    {'name': 'Eternity', 'fear': 61, 'status': 'contained'},
    {'name': 'Tomato',   'fear':  3, 'status': 'at large'},
]

for d in devils:
    print(f"{d['name']:10s} fear {d['fear']:3d}   {d['status']}")
```

**A list of dictionaries is the shape of almost all real data.** A spreadsheet
is a list of dictionaries. A database table is a list of dictionaries. The
contacts on your phone are a list of dictionaries.

Learn to work with this one shape and you can handle most data you'll ever meet.

(That `:10s` and `:3d` in the f-string just lines the columns up — 10 characters
wide for text, 3 for the number. Handy, not important.)


## Have we seen it before?

A hunter radios in a name. Is it already on file? The keyword is `in`:


```python
known = []
for d in devils:
    known.append(d['name'])

print(known)
print()
print('Gun in the archive?     ', 'Gun' in known)
print('Chainsaw in the archive?', 'Chainsaw' in known)
```

## Only the ones that matter

Nobody wants the whole list. They want *the dangerous ones that are still out
there.* Two conditions at once, joined with `and`:


```python
for d in devils:
    if d['fear'] > 70 and d['status'] == 'at large':
        print(f"PRIORITY — the {d['name']} Devil (fear {d['fear']})")
```

There are three of these joining words and they do what they sound like:

| | |
|---|---|
| `and` | both have to be true |
| `or` | either one will do |
| `not` | flips it |

## Reporting to Makima

She doesn't want the list, she wants the numbers. Four built-in tools do almost
all of it:


```python
fears = []
for d in devils:
    fears.append(d['fear'])

print('devils on file: ', len(devils))          # how many
print('highest fear:   ', max(fears))           # biggest
print('lowest fear:    ', min(fears))           # smallest
print('total fear:     ', sum(fears))           # added up
print('average fear:   ', sum(fears) / len(fears))
```

That average is ugly — `63.8` would do. `round` takes a second number saying how
many decimal places you want:


```python
average = sum(fears) / len(fears)

print(round(average, 1))
print(round(average))
```

## Ranking them

`sorted` puts things in order. On plain numbers it just works:


```python
print(sorted(fears))
print(sorted(fears, reverse=True))     # biggest first
```

But we don't want a list of loose numbers, we want the *records* in order. So
you have to tell `sorted` **what to sort by** — that's what `key=` is for:


```python
by_threat = sorted(devils, key=lambda d: d['fear'], reverse=True)

for rank, d in enumerate(by_threat, start=1):
    print(f"{rank}. {d['name']:10s} {d['fear']:3d}")
```

`lambda d: d['fear']` looks strange but says something simple: *for each record,
the thing to sort by is its fear.* You'll write hundreds of these eventually.

## What's inside `max`

`max` is useful, but it's worth knowing it isn't magic. Here's how you'd find
the worst devil without it:


```python
worst = devils[0]              # assume the first one is the worst

for d in devils:
    if d['fear'] > worst['fear']:
        worst = d              # found something worse? that's the new worst

print(f"most dangerous on file: the {worst['name']} Devil")
```

Read that loop again, because the shape of it comes up constantly:

> **Keep a 'best so far'. Walk through everything. Replace it when you find
> something better.**

That's how you find a maximum, a minimum, the closest thing, the cheapest
flight, the shortest route. It's also — almost exactly — how machine learning
works, which is a conversation for another day.

## Power's paperwork

Time for the hunters. Denji, Power and Aki file their own kill counts, which is
a mistake Public Safety will come to regret.


```python
hunters = [
    {'name': 'Denji', 'kills': 12},
    {'name': 'Power', 'kills': 'a million'},
    {'name': 'Aki',   'kills': 40},
]

for h in hunters:
    print(f"{h['name']:6s} {h['kills']}")
```

Looks fine. It is not fine.

Every value in Python has a **type** — what kind of thing it is. `type()` tells
you:


```python
print(type(12))
print(type('a million'))
print(type(3.5))
print(type(True))
print()
print('Denji:', type(hunters[0]['kills']))
print('Power:', type(hunters[1]['kills']))
```

`int` is a whole number. `str` is text — a *string* of characters. `float` has a
decimal point. `bool` is True or False.

Denji's kill count is a number. **Power's is text.** She typed words into a
number field, which Python was perfectly happy to let her do.

Now watch what happens when Makima asks for the division total.

**This next cell is meant to fail.** Run it and read the last line.


```python
total = 0

for h in hunters:
    total = total + h['kills']

print(total)
```

> `TypeError: unsupported operand type(s) for +: 'int' and 'str'`

*I can't add a number and some text.* And it's right — what would `12 + 'a
million'` even be?

**This is one of the most common bugs in real software.** Somewhere a form let
someone type words where a number belonged, and everything downstream fell over.

You can convert between types when it makes sense:


```python
print(int('12') + 5)                 # text that looks like a number -> number
print(str(12) + ' devils')           # number -> text, so it can join a sentence
print(float('3.5') * 2)

print()
print('but only when it makes sense:')
print(int('a million'))              # this one can't work
```

`ValueError: invalid literal for int() with base 10: 'a million'`

A different error for a different problem. `TypeError` means *wrong kind of
thing*. `ValueError` means *right kind of thing, nonsense contents*.

So the real fix isn't converting — it's **checking before you trust it**:


```python
total = 0

for h in hunters:
    if type(h['kills']) != int:
        print(f"REJECTED — {h['name']} filed '{h['kills']}', which is not a number")
    else:
        total = total + h['kills']

print()
print('verified division total:', total)
```

That's called **validating your input**, and it is the difference between
software that survives contact with real people and software that doesn't.

Power still isn't getting credit for a million kills.

## Reports from the field

Hunters radio in names while running. What reaches HQ is a mess:


```python
reports = ['  gun devil ', 'BOMB DEVIL', 'katana Devil  ',
           "   makima's devil"]

for r in reports:
    print(repr(r))
```

`repr()` shows you the string *including* its quotes and spaces, which is how
you spot the stray whitespace that's about to ruin your day.

Strings come with tools to clean themselves up. You call them with a dot:


```python
messy = '  gun devil '

print(repr(messy.strip()))        # cut whitespace off both ends
print(messy.strip().upper())      # SHOUTING
print(messy.strip().lower())      # quiet
print(messy.strip().title())      # Capital Letters On Each Word
print(messy.strip().replace('devil', 'Devil'))
```

Notice you can chain them: `.strip().title()` cleans, then capitalizes. Each one
hands its result to the next.

So the whole cleanup is one line inside a loop:


```python
for r in reports:
    print(f'{r!r:20}  ->  {r.strip().title()}')
```

**Your turn.** Run the cleanup on all four and look at each result
carefully. Three of them come out perfect.

The last one comes out as `Makima'S Devil`.

**Why?** `.title()` capitalizes the first letter of every word — and it thinks
an apostrophe starts a new word. Nobody wrote a bug. The method did exactly
what it does; it just doesn't match what we meant by "word".

This happens constantly with built-in tools. They're not wrong, they're just
narrower than you assumed. **Look at your output instead of trusting it.**

## One more trap

Ask a dictionary for a label it doesn't have and it will tell you off. Run this:


```python
print(devils[0]['weakness'])
```

`KeyError: 'weakness'` — *there's no such label in this record.*

It's a good error, because the alternative is worse. A language that quietly
returned nothing here would let you build a whole report out of blanks and never
mention it.

If you're not sure a label exists, `.get()` asks politely instead:


```python
print(devils[0].get('weakness'))                    # None - no such label
print(devils[0].get('weakness', 'not recorded'))     # or say what to use instead
print(devils[0].get('name', 'not recorded'))         # this one does exist
```
