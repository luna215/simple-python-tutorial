# Blueprints

Last time you built Public Safety's archive out of dictionaries. It worked.

It's also held together with tape, and here's why.


```python
gun    = {'name': 'Gun',    'fear': 99, 'status': 'at large'}
bomb   = {'name': 'Bomb',   'fear': 82}
katana = {'name': 'Katana', 'fear': 74, 'staus': 'contained'}

for d in [gun, bomb, katana]:
    print(d)
```

Read those three carefully.

**Bomb has no `status`.** Somebody forgot it.

**Katana has `staus`.** Somebody typed it wrong, so it has a field nobody will
ever look for and no field anybody will.

Python was perfectly happy with all three. No error, no warning, no hint. The
trouble arrives an hour later, somewhere else, when a report comes out blank and
you have no idea why.

Nothing in that code says *what a devil is*. Every record is a fresh guess.

So let's write down what a devil is, once.


---

## The blueprint

A **class** is how we describe objects in a computer. Computers are dumb. They need explicit instructions and description.
The class below tells the computer: "This is what a devil is".


```python
class Devil:

    def __init__(self, name, fear):
        self.name = name
        self.fear = fear
        self.status = 'at large'
```

Four new things in six lines. Let's take them one at a time.

**`class Devil:`** — everything indented under this describes what a devil is.
By tradition class names start with a Capital Letter.

**`__init__`** — those are two underscores on each side. It's the *setup*
function, and Python runs it automatically every time you make a new devil.
Say it out loud as "init", short for initialise.

**`self`** — the devil currently being built. Not devils in general — *this one*.

**`self.name = name`** — take the name that was handed in, and stick it onto
this devil. From now on that devil carries its own name around with it.

Nothing happened when you ran that cell. You've described a devil, not made one.


## Making them

Now use the blueprint. You don't call `__init__` yourself — writing `Devil(...)`
does it for you:


```python
gun = Devil('Gun', 99)
tomato = Devil('Tomato', 3)

print(gun.name, gun.fear, gun.status)
print(tomato.name, tomato.fear, tomato.status)
```

Two devils, from one blueprint. Notice the dot: `gun.name` instead of
`gun['name']`. Slightly less typing, and it will matter more in a minute.

**Notice what you didn't have to do.** You never wrote `'status'` when making
either of them, but both have one — the blueprint filled it in. Every devil that
comes out of this mould has all three fields. Every single time. That's the
guarantee dictionaries could never give you.


### And if you forget something?

This is the part worth caring about. **This cell is meant to fail** — run it and
read the message.


```python
bomb = Devil('Bomb')
```

It tells you the exact thing you left out, at the exact moment you did it —
not an hour later in a blank report.

Compare that with `{'name': 'Bomb', 'fear': 82}` from the top of this page,
which sailed through without a murmur.

**That is what a class buys you.** Not tidier code. A promise that every devil
is the same shape.

## Devils that do things

So far a devil just sits there holding facts. Dictionaries could do that.

The real change is that a class can hold **things a devil can do**. A function
living inside a class is called a **method**, and its first parameter is always
`self` — the devil it's being asked about.


```python
class Devil:

    def __init__(self, name, fear):
        self.name = name
        self.fear = fear
        self.status = 'at large'

    def report(self):
        return f'The {self.name} Devil — fear {self.fear} — {self.status}'

    def contain(self):
        self.status = 'contained'

    def killed(self):
        self.status = 'reincarnating in hell'


gun = Devil('Gun', 99)
print(gun.report())
```

`report` doesn't take any information in. It doesn't need to — it's *attached to
a devil*, and `self` is that devil. `self.name` is this devil's name.

Now watch a devil change:


```python
gun = Devil('Gun', 99)
print(gun.report())

gun.contain()
print(gun.report())

gun.killed()
print(gun.report())
```

`contain()` changed something *inside* `gun`, and `gun` remembered. That's the
whole idea of an object: a thing that keeps its own state and knows how to
change it.

(Devils reincarnating in hell when they die is straight out of the manga. It's
also a perfectly good method.)

## Making `print` behave

Try printing a devil directly, without `.report()`:


```python
print(gun)
```

Ugly, and not useful. Python doesn't know how you'd like a devil written out —
so tell it. `__str__` is the method Python looks for whenever something needs to
become text:


```python
class Devil:

    def __init__(self, name, fear):
        self.name = name
        self.fear = fear
        self.status = 'at large'

    def __str__(self):
        return f'The {self.name} Devil — fear {self.fear} — {self.status}'

    def contain(self):
        self.status = 'contained'


gun = Devil('Gun', 99)
print(gun)
print(f'Priority target: {gun}')
```

`print` and f-strings both use it, for free. Those double-underscore names are
hooks Python already knows to look for — `__init__` for setup, `__str__` for
text. There are dozens more, and you've now met the two that matter most.


## You have invented a type

Last lesson you met `int`, `str`, `float` and `bool`. Look at what `type()` says
about a devil:


```python
print(type(99))
print(type('Gun'))
print(type(gun))
```

`Devil` sits in that list as an equal. You didn't use a type — you **made** one.
It didn't exist an hour ago.

That's what people mean when they say a language is *object-oriented*: you get
to add your own nouns to it.

## Every devil is a devil. No two devils are alike.

In *Chainsaw Man* every devil follows the same rules — named after a fear, dies
and reincarnates, can make contracts. But the Gun Devil does something the Bomb
Devil doesn't.

You could copy the whole `Devil` class and change one method. Don't. Instead say
*this new thing is a Devil, plus a difference* — which is written
`class GunDevil(Devil):`


```python
class Devil:
    def __init__(self, name, fear):
        self.name = name
        self.fear = fear
        self.status = 'at large'

    def __str__(self):
        return f'The {self.name} Devil — fear {self.fear}'

    def power(self):
        return 'nothing anyone has written down'


class GunDevil(Devil):
    def power(self):
        return 'kills a thousand people in five seconds'


class BombDevil(Devil):
    def power(self):
        return 'detonates whatever it touches'


for d in [GunDevil('Gun', 99), BombDevil('Bomb', 82), Devil('Tomato', 3)]:
    print(f'{d}')
    print(f'   -> {d.power()}')
```

Look at what `GunDevil` did **not** contain: no `__init__`, no `__str__`. It
inherited both from `Devil` and only wrote down its difference.

That's **inheritance**: `Devil` is the *parent*, `GunDevil` is the *child*, and
the child gets everything the parent had unless it says otherwise.

When a child does say otherwise — `GunDevil` writing its own `power` — that's
called **overriding**. Python uses the child's version.

## Is this thing a devil?

`isinstance` answers that, and it counts children too:


```python
gun = GunDevil('Gun', 99)

print('gun is a GunDevil:', isinstance(gun, GunDevil))
print('gun is a Devil:   ', isinstance(gun, Devil))
print('a Devil is a GunDevil:', isinstance(Devil('Tomato', 3), GunDevil))
```

A Gun Devil **is** a Devil. A plain Devil is not a Gun Devil. The relationship
only runs one way, which is exactly how it works in the manga too.

## Break it on purpose

**These four tell you off.** One at a time; read the message; undo it.


```python
gun = GunDevil('Gun', 99)

print(gun.nmae)          # a field that does not exist
```

```python
Devil('Gun', 99, 'at large')     # one argument too many
```

```python
class Broken(Devil):
    def shout():                  # forgot self
        return 'RAAAA'

Broken('Noise', 40).shout()
```

```python
class AlsoBroken:
    def __init__(self, name):
        name = name               # forgot self. — sets a local, not the devil

AlsoBroken('Gun').name
```

That last one is the sneakiest of the four. `name = name` is a perfectly legal
line that quietly does nothing useful — it sets a variable that vanishes the
moment `__init__` ends. The devil never gets a name at all.

**Forgetting `self.` is the single most common mistake people make with classes.**
You will do it. Now you'll recognise it.


### And one that doesn't tell you off

Classes are a promise about what a devil *starts with*. They are not a guard on
everything you do afterwards:


```python
gun = GunDevil('Gun', 99)

gun.fera = 100          # meant fear, typed fera

print('fear:', gun.fear)
print('and now there is also:', gun.fera)
```

No error. Python let you bolt a brand new field onto that one devil, exactly
like the dictionary typo at the top of this page.

So be honest about what you gained: **`__init__` guarantees every devil starts
the same shape, and methods keep the behaviour next to the data.** It doesn't
babysit you after that. Nothing does.

|---|---|
| `class Devil:` | a blueprint — the idea of a thing |

## Why this is the big one

Almost everything you'll ever use in Python is an object made from a class
somebody wrote. A string is. A list is. `.append` and `.strip` are methods,
exactly like `.contain()` — someone wrote a `list` class and gave it an
`append` method, and you have been using their blueprint all along.

You just crossed from using other people's nouns to writing your own.
