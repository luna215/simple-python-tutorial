## One name, many values

A normal variable holds **one** thing:


```python
x = 200

print(x)
```

A **list** holds a whole row of them. Square brackets, commas in between:


```python
crowd = [100, 250, 400, 550, 700]

print(crowd)
```

Five numbers, one name.

`len()` counts how many are in there:


```python
print(len(crowd))
```

## The `for` loop

A `for` loop does something **once for each thing in the list**:


```python
for x in crowd:
    print('someone is standing at', x)
```

Read it as English: *for each x in the crowd, print where they are.*

Python ran that indented line five times. First `x` was 100, then 250, then 400,
550, 700 — then the list ran out and it stopped.

Notice you never told it *how many* times to go round. It just uses however many
are in the list.

## Building a list with a loop

Typing out a hundred numbers by hand would be miserable. Two tools fix that.

**`range(n)`** counts from 0 up to *but not including* n. That catches everyone
out once, so look carefully at what this prints:


```python
print(list(range(5)))
```

Five numbers — starting at **0**, not 1, and stopping at **4**, not 5.

**`.append(...)`** sticks one more thing on the end of a list:


```python
crowd = []          # an empty list - nothing in it yet

crowd.append(100)
crowd.append(250)

print(crowd)
```

Put those together and you can build a list of any size:


```python
crowd = []

for i in range(20):
    crowd.append(i * 50)

print(crowd)
print(len(crowd), 'people')
```

**Try it:** change `i * 50` to `i * 20`. Then to `i * 100`.

What does that number actually control?

## The trap

This one catches everybody at least once, so let's get it over with now.

Here is a loop that looks like it moves everyone 200 to the right.

**Predict what the last line prints before you run it.**


```python
crowd = [100, 250, 400]

for x in crowd:
    x = x + 200

print(crowd)
```

Nothing moved.

`x` is a **copy** of each number, handed to you fresh each time round the loop.
You can change the copy all you like — the list never hears about it.

There was no error. No warning. Python did exactly what you asked, and what you
asked wasn't what you meant.

Those are the expensive bugs: the ones that don't announce themselves. Getting
used to the idea now will save you an afternoon later.
