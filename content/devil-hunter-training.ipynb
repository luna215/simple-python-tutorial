```python
name = 'Chainsaw'
fear = 95

print(name, fear)
```

```python
name = 'Chainsaw'
fear = 95

print(f'The {name} Devil — fear level {fear}')
```

## Making the number mean something

Right now `fear` is just sitting there. Let's make Public Safety actually
respond to it.

`if` runs a line **only when something is true**. `elif` means "or else, if…"
and `else` means "or else, in every other case":


```python
name = 'Chainsaw'
fear = 95

if fear > 80:
    print(f'The {name} Devil — EVACUATE THE CITY')
elif fear > 40:
    print(f'The {name} Devil — send a hunter')
else:
    print(f'The {name} Devil — ignore it, it will be fine')
```

The indented lines are the ones that only run sometimes. Python uses the
**spaces at the start of a line** to know what belongs to what. Other languages
use curly brackets. Python uses the shape of the page.

## Stop copy-pasting

To make a second devil right now, you'd copy all nine lines and change two
words. Do that five times and your code is a mess.

Devil hunters have paperwork for repeated jobs. Programmers have **functions**.

`def` means "define". The words in the brackets are the blanks you fill in
each time you use it:


```python
def devil(name, fear):
    if fear > 80:
        threat = 'EVACUATE THE CITY'
    elif fear > 40:
        threat = 'send a hunter'
    else:
        threat = 'ignore it, it will be fine'

    print(f'The {name} Devil — fear {fear} — {threat}')


devil('Gun', 99)
devil('Bomb', 70)
devil('Tomato', 3)
```

Three devils, three lines. The nine lines of logic got written **once**.

That's the entire point of functions, and it's most of what separates code that
grows from code that collapses.

## An entire outbreak

A **list** holds many things under one name. Square brackets, commas between.

A `for` loop then does something **once for each thing in the list**:


```python
import random

nouns = ['Gun', 'Bomb', 'Future', 'Tomato', 'Darkness',
         'Homework', 'Monday', 'Dentist', 'Wifi', 'Spider']

for name in nouns:
    devil(name, random.randint(1, 100))
```

Ten devils. You wrote the word `devil` once.

`random.randint(1, 100)` picks a whole number between 1 and 100, fresh each
time — so **run that cell again**. Different outbreak. The Dentist Devil might
be a citywide emergency this time.

## Contracts

Nobody in *Chainsaw Man* gets power for free. You make a **contract**: you give
something up, and you get something back. Usually a terrible deal. Denji trades
away rather a lot.

Here's the thing that makes this worth explaining:

> **A contract is exactly what a function is.** You hand it something. It hands
> something back.

The new word is `return` — it means *hand this back to whoever asked*:


```python
def contract(offer):
    if offer == 'your eye':
        return 'you can see two minutes into the future'
    if offer == 'your years':
        return 'the Curse Devil answers'
    if offer == 'your heart':
        return 'you become Chainsaw Man'
    return 'the devil laughs at you'


print(contract('your eye'))
print(contract('your heart'))
print(contract('a ham sandwich'))
```

Note `==` with **two** equals signs. One `=` means *make this equal that*. Two
means *is this equal to that?* Mixing them up is a rite of passage.

### `print` and `return` are not the same thing

This trips up everybody, so let's be clear about it.

`print` **shows you** something and then it's gone. `return` **hands it back**
so you can keep it and use it later:


```python
power = contract('your heart')

print(f'Denji gave up his heart.')
print(f'In exchange: {power}')
print(f'That deal is {len(power)} characters long, which feels short.')
```

The `power` variable is holding the thing the function handed back. You can't
do that with something you only printed.

## Denji's actual goals

The joke that makes Chainsaw Man work is that Denji has world-ending power and
his ambitions are microscopic. He wants bread with jam on it. He wants to sleep
somewhere warm.

Lists again — and notice the loop doesn't care how many there are:


```python
dreams = ['eat bread with jam on it',
          'sleep in a real bed',
          'have someone say his name nicely']

for number, dream in enumerate(dreams, start=1):
    print(f'Goal {number}: {dream}')
```

`enumerate` hands you the position **and** the item, so you get numbering for
free. `start=1` because computers count from 0 and humans don't.

the loop — that's the point of loops.

## The fight

Everything so far ran a fixed number of times. A fight doesn't work like that —
it goes on **until somebody drops**.

That's a `while` loop: *keep going as long as this stays true.*

Read it, run it. Then run it again. And again.


```python
import random

denji = 100
devil_hp = 100
devil_name = 'Katana'
round_number = 1

while denji > 0 and devil_hp > 0:
    hit = random.randint(8, 30)
    devil_hp = devil_hp - hit
    print(f'Round {round_number}: Denji rips for {hit}. '
          f'{devil_name} Devil at {max(devil_hp, 0)} HP')

    if devil_hp <= 0:
        break

    bite = random.randint(8, 30)
    denji = denji - bite
    print(f'          {devil_name} Devil bites for {bite}. '
          f'Denji at {max(denji, 0)} HP')

    round_number = round_number + 1

print()
if denji > 0:
    print(f'Denji wins with {denji} HP left.')
else:
    print('Denji is down. Pochita is not happy.')
```

**Run it five times.** Denji wins about two times in three — so losing is
real, not decoration. You wrote one fight and got a different one every time.

`break` means *leave the loop right now* — without it the devil gets a free bite
in after it's already dead.

## One more thing, from the same author

*Look Back* is by Tatsuki Fujimoto too, and it is a very different kind of story.

Fujino decides she's going to get good at drawing. There's no montage, no
shortcut, no devil to make a contract with. She just draws. Every day. For years.
Fujimoto gives you a page of the same desk while the seasons change outside the
window.

Here is that page, in Python.


```python
panels = 0

for day in range(365):
    panels = panels + 4        # four panels a day. that's all.

print(f'{panels} panels in one year')
```

Four panels. Every day. One year.

Now the part that's genuinely hard to believe. Suppose each day you don't get
*better* by much — just **one percent** better than you were yesterday:


```python
skill = 1

for day in range(365):
    skill = skill * 1.01

print(f'After one year you are {round(skill)}x better')
```

Thirty-eight times.

Not because any single day mattered. Because the loop ran 365 times.

Let's watch four years go by. This is a loop **inside** a loop — the inner one
runs a whole year, the outer one counts the years:


```python
skill = 1

for year in range(1, 5):
    for day in range(365):
        skill = skill * 1.01

    print(f'After year {year}: {round(skill)}x better')
```

You wrote six lines. Python ran them 1,460 times.

That's the whole reason loops exist — and, more or less, the whole reason Fujino
kept sitting at that desk.

## What you just learned

Seriously — this is not a small list:

| | |
|---|---|
| `name = 'Chainsaw'` | a **variable** — a fact with a name |
| `f'The {name} Devil'` | an **f-string** — drop variables into text |
| `if` / `elif` / `else` | do something **only sometimes** |
| indentation | how Python knows what's inside what |
| `def devil(name, fear):` | a **function** — write it once, use it forever |
| `return` | hand a result **back** (not the same as `print`) |
| `==` | *is this equal to?* (one `=` means *make it equal*) |
| `['Gun', 'Bomb']` | a **list** — many things, one name |
| `for name in nouns:` | do something **once for each** |
| `while denji > 0:` | keep going **until** something changes |
| `break` | leave the loop right now |
| `range(5)` | 0, 1, 2, 3, 4 — starts at zero, stops one early |
| `random.randint(1, 100)` | a number nobody chose |
| `enumerate` | position and item together |

That's most of the toolkit. Everything after this is combinations of these six
ideas.
